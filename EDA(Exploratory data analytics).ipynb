# 3. we can fill the mean /median/mode in place of null.

```python
# filling mean value in place null

# Interview : when to use mean to replace the null value
# when the data is continuous or discrete (Numeric).
# Don't use the mean to replace Null value if dataset is having lots of outliers

# why mean is highly sensitive to outliers 
# Mean is highly sensitive to outliers because it is calculated from the sum of all values. 
#An extreme value greatly changes the total sum, causing the average to shift significantly.

# why we Don't use the mean to replace Null value if dataset is having lots of outliers 
# You should not use the mean to replace missing (null) values
#when the dataset has many outliers because the mean is highly sensitive to extreme values.

# If the dataset is normal distributed
# using the mean to replace missing values is generally a good choice.
```

```python
short_data = [np.nan,15,20,50,70,90,40,np.nan,np.nan,0,np.nan]

df4 = pd.DataFrame(short_data)
df4
```

# Fill the NULL value with bfill & ffill

```python
# bfill: backforward fill : propogate the next valid value backward to fill missing value -- if last value is null it will stay null
# ffill: forward fill: propogate the previous valid value forward to fill missing value -- the previous value will be filled
```

```python
fav_drinks = [np.nan, 'water' , 'chocplate drink' ,np.nan, np.nan, np.nan,  'coconut', np.nan,  'monstor', 'black dog', np.nan, 'sprit', 
              'thumbsup', np.nan]
```

```python
df3 = pd.DataFrame(fav_drinks, columns = ['drinks'])
df3
```

```python
# backward fill(bfill)

df3 = df3.bfill()
df3
```

```python
fav_drinks = [np.nan, 'water' , 'chocplate drink' ,np.nan, np.nan, np.nan,  'coconut', np.nan,  'monstor', 'black dog', np.nan, 'sprit', 
              'thumbsup', np.nan]
df3 = pd.DataFrame(fav_drinks, columns = ['drinks'])
df3
```

```python
# forward fill(f-fill)
df3 = df3.ffill()
df3
```

# IQR (Interquantile range)

```python
Salary = [10, 20, 30, 40, 50, 60, 70, 80, 100, 200]
sns.boxplot(x=Salary)
```

```python
Salary = [10, 20, 30, 40, 50, 60, 70, 80, 100, 200]

df1 = pd.DataFrame(Salary, columns = ['Salary'])

# Q1 = 25th percentile => 25% of data will be below to Q1
Q1 = df1.Salary.quantile(0.25)
Q3 = df1.Salary.quantile(0.75)
IQR = Q3 - Q1
print(IQR)
```

```python
upper_bound = Q3 + 1.5 * IQR
lower_bound = Q1 - 1.5 * IQR

print(upper_bound)
print(lower_bound)
```

# identify and Handle the Outliers

```python
# The data points which are smaller than -3 sd and greater than +3 sd
# z_score = how much sd the data points are far away from the mean
# z-score = (data - mean)/sd

df1 = df.copy()
df1['z-score'] = (df['age'] - df['age'].mean())/df['age'].std()
outliers = df1[(df1['z-score'] < -3) | (df1['z-score'] < 3)]

df1[(df1['z-score'] >= -3) & (df1['z-score'] <= 3)]
df1
```
